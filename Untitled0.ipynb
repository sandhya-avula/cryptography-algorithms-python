```python

from cryptography.hazmat.primitives.ciphers.aead import AESGCM
import os
import base64

# Generate a random 256-bit AES key
key = AESGCM.generate_key(bit_length=256)
aes = AESGCM(key)

# Sample message
message = "Hello! This is my cybersecurity internship project."

# Generate a unique 12-byte nonce for this encryption
nonce = os.urandom(12)

# Encrypt the message
ciphertext = aes.encrypt(nonce, message.encode("utf-8"), None)

# Encode encrypted data for readable display
encrypted_text = base64.b64encode(ciphertext).decode("utf-8")

# Decrypt the message
decrypted_message = aes.decrypt(
    nonce, ciphertext, None
).decode("utf-8")

print("Original message:", message)
print("Encrypted message:", encrypted_text)
print("Decrypted message:", decrypted_message)
print("Decryption successful:", message == decrypted_message)
```

```python

import hashlib

message1 = "Hello Cybersecurity"
message2 = "Hello cybersecurity"

hash1 = hashlib.sha256(message1.encode("utf-8")).hexdigest()
hash2 = hashlib.sha256(message2.encode("utf-8")).hexdigest()

print("Message 1:", message1)
print("SHA-256:", hash1)

print("\nMessage 2:", message2)
print("SHA-256:", hash2)

print("\nHashes are different:", hash1 != hash2)
print("Hash length:", len(hash1), "characters")

from cryptography.exceptions import InvalidTag

# Test 1: Correct key and ciphertext
assert aes.decrypt(nonce, ciphertext, None).decode() == message
print("PASS: Correct ciphertext decrypts successfully.")

# Test 2: Modified ciphertext
tampered = bytearray(ciphertext)
tampered[0] ^= 1

try:
    aes.decrypt(nonce, bytes(tampered), None)
    print("FAIL: Modified ciphertext was accepted.")
except InvalidTag:
    print("PASS: Modified ciphertext was rejected.")

# Test 3: SHA-256 detects a message change
assert hashlib.sha256(b"hello").digest() != hashlib.sha256(b"Hello").digest()
print("PASS: Changed message produces a different hash.")
```
